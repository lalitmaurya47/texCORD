```python
import os
import cv2
from pathlib import Path
from tqdm import tqdm


def extract_ultrasound_region(img):
    """
    Crop ultrasound-only region from the image.
    Adjust coordinates if needed for your dataset.
    """
    return img[33:451, 98:584]


def process_dataset(input_dir, output_dir):
    input_dir = Path(input_dir)
    output_dir = Path(output_dir)

    # Find all PNG files recursively
    png_files = list(input_dir.rglob("*.png"))

    print(f"Found {len(png_files)} PNG files")

    for file_path in tqdm(png_files, desc="Processing images"):

        # Read image
        img = cv2.imread(str(file_path))

        if img is None:
            print(f"Could not read: {file_path}")
            continue

        # Extract ultrasound region
        cropped = extract_ultrasound_region(img)

        # Preserve folder structure
        relative_path = file_path.relative_to(input_dir)
        save_path = output_dir / relative_path

        # Create output subfolder if needed
        save_path.parent.mkdir(parents=True, exist_ok=True)

        # Save cropped image
        cv2.imwrite(str(save_path), cropped)

    print("Processing complete.")


# Example usage
input_folder = "/mnt/lustre/lalit47/us_data/pre_train/JoCoHS/data/image"
output_folder = "/mnt/lustre/lalit47/us_data/pre_train/JoCoHS_new/data/image"

process_dataset(input_folder, output_folder)
```
