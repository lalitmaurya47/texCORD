```python
import os
import cv2
import shutil

# =========================
# USER INPUT PATHS
# =========================

source_path = r"/mnt/lustre/lalit47/us_data/bench/uterine fibroid ultrasound images/datasets"
target_path = r"/mnt/lustre/lalit47/down/class/Uterine"

image_size = (256, 256)

splits = ["train", "test"]
class_names = ["NUF", "UF"]


def process_and_save(src, dst):
    im = cv2.imread(src, cv2.IMREAD_GRAYSCALE)

    if im is None:
        print(f"Failed to read: {src}")
        return False

    im = cv2.resize(im, image_size, interpolation=cv2.INTER_LINEAR)

    os.makedirs(os.path.dirname(dst), exist_ok=True)
    cv2.imwrite(dst, im)

    return True


def preprocess_split(input_split, output_split):
    """
    input_split: train or test folder from source dataset
    output_split: train, val, or test folder in target dataset
    """

    for class_name in class_names:
        input_dir = os.path.join(source_path, input_split, class_name)
        output_dir = os.path.join(target_path, output_split, class_name)

        if not os.path.isdir(input_dir):
            print(f"Missing folder: {input_dir}")
            continue

        os.makedirs(output_dir, exist_ok=True)

        count = 0

        for file_name in os.listdir(input_dir):
            if not file_name.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff")):
                continue

            src_img = os.path.join(input_dir, file_name)

            base_name = os.path.splitext(file_name)[0]
            dst_img = os.path.join(output_dir, base_name + ".png")

            success = process_and_save(src_img, dst_img)

            if success:
                count += 1

        print(f"{output_split}/{class_name}: {count} images saved")


# =========================
# CREATE TRAIN
# =========================

preprocess_split(input_split="train", output_split="train")

# =========================
# CREATE TEST
# =========================

preprocess_split(input_split="test", output_split="test")

# =========================
# CREATE VAL SAME AS TEST
# =========================

preprocess_split(input_split="test", output_split="val")

print("\nClassification dataset created successfully!")
print("Validation set is the same as test set.")
```
